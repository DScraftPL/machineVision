```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("OpenCV:", cv2.__version__)
print("NumPy:", np.__version__)
```

```python
obraz = cv2.imread("obraz_testowy.jpg")
if obraz is None:
    print("Nie udalo sie wczytac obrazu. Sprawdz nazwe pliku.")
else:
    print(type(obraz))
print("Wymiary:", obraz.shape)
print("Typ danych:", obraz.dtype)
```

```python
plt.imshow(obraz)
plt.axis("off")
plt.show()
```

```python
obraz_rgb = cv2.cvtColor(obraz, cv2.COLOR_BGR2RGB)
plt.imshow(obraz_rgb)
plt.axis("off")
plt.show()
```

```python
piksel = obraz_rgb[50, 100]
print("Wartosci RGB piksela:", piksel)
```

```python
obraz_rgb[0:60, 0:60] = [255, 0, 0]
plt.imshow(obraz_rgb)
plt.axis("off")
plt.show()
```

```python
fragment = obraz_rgb[100:300, 200:450]
plt.imshow(fragment)
plt.axis("off")
plt.show()
```

```python
kanal_r = obraz_rgb[:, :, 0]
kanal_g = obraz_rgb[:, :, 1]
kanal_b = obraz_rgb[:, :, 2]
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(kanal_r, cmap="gray")
plt.title("Kanal R")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(kanal_g, cmap="gray")
plt.title("Kanal G")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(kanal_b, cmap="gray")
plt.title("Kanal B")
plt.axis("off")
plt.show()
```

```python
obraz_szary = cv2.cvtColor(obraz, cv2.COLOR_BGR2GRAY)
print("Wymiary obrazu szarego:", obraz_szary.shape)
plt.imshow(obraz_szary, cmap="gray")
plt.axis("off")
plt.show()
cv2.imwrite("wynik_szary.jpg", obraz_szary)
```

# ZADANIE 4

```python
kanal_r = obraz[:, :, 0]
kanal_g = obraz[:, :, 1]
kanal_b = obraz[:, :, 2]

obraz_copy = obraz.copy()

obraz_copy[:, :, 0] = kanal_b
obraz_copy[:, :, 2] = kanal_r

plt.imshow(obraz)
plt.axis("off")
plt.show()

plt.imshow(obraz_copy)
plt.axis("off")
plt.show()
```

Kolory zostały zamienione ze sobą, wysokości jednego koloru zostały przypisane do innego, co daje efekt zamiany kolorów.

# ZADANIE DODATKOWE

```python
obraz = cv2.cvtColor(cv2.imread("obraz_testowy.jpg"), cv2.COLOR_BGR2RGB)
if obraz is None:
    print("Nie udalo sie wczytac obrazu. Sprawdz nazwe pliku.")

print(
    f"Obraz ma {obraz.shape[0]} wierszy, {obraz.shape[1]} kolumn, {obraz.shape[2]} kanały barwne"
)

plt.imshow(obraz)
plt.axis("off")
plt.show()

kanal_r = obraz[:, :, 0]
kanal_g = obraz[:, :, 1]
kanal_b = obraz[:, :, 2]

obraz[:, :, 0] = 255 - kanal_r
obraz[:, :, 1] = 255 - kanal_g
obraz[:, :, 2] = 255 - kanal_b

plt.imshow(obraz)
plt.axis("off")
plt.show()

obraz = cv2.imread("obraz_testowy.jpg")
if obraz is None:
    print("Nie udalo sie wczytac obrazu. Sprawdz nazwe pliku.")

obraz = cv2.cvtColor(obraz, cv2.COLOR_BGR2RGB)

obraz[:, :, 1] = 0
obraz[:, :, 2] = 0

plt.imshow(obraz)
plt.axis("off")
plt.show()
```
