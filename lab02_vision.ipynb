```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

obraz = cv2.imread("obraz_testowy.jpg")
obraz_rgb = cv2.cvtColor(obraz, cv2.COLOR_BGR2RGB)
plt.imshow(obraz_rgb)
plt.axis("off")
plt.show()
```

```python
proba = obraz_rgb + 50
plt.imshow(proba)
plt.axis("off")
plt.show()
```

```python
jasniejszy = np.clip(obraz_rgb.astype(np.int16) + 50, 0, 255).astype(np.uint8)
ciemniejszy = np.clip(obraz_rgb.astype(np.int16) - 50, 0, 255).astype(np.uint8)
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(ciemniejszy)
plt.title("Ciemniejszy")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(obraz_rgb)
plt.title("Oryginal")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(jasniejszy)
plt.title("Jasniejszy")
plt.axis("off")
plt.show()
```

```python
wiekszy_kontrast = cv2.convertScaleAbs(obraz_rgb, alpha=1.6, beta=0)
mniejszy_kontrast = cv2.convertScaleAbs(obraz_rgb, alpha=0.5, beta=0)
negatyw = 255 - obraz_rgb
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(obraz_rgb)
plt.title("Oryginal")
plt.axis("off")
plt.subplot(2, 2, 2)
plt.imshow(negatyw)
plt.title("Negatyw")
plt.axis("off")
plt.subplot(2, 2, 3)
plt.imshow(wiekszy_kontrast)
plt.title("Wiekszy kontrast")
plt.axis("off")
plt.subplot(2, 2, 4)
plt.imshow(mniejszy_kontrast)
plt.title("Mniejszy kontrast")
plt.axis("off")
plt.show()
```

```python
# Konwersja do typu float32
obraz_float = obraz_rgb.astype(np.float32)
# centrowanie wokół 0
obraz_centrowany = obraz_float - 127.5
# zmiana kontrastu
alpha_nowy_kontrast = 1.6
obraz_skalowany = obraz_centrowany * alpha_nowy_kontrast
# powrót do zakresu 0 255
obraz_przesuniety_z_powrotem = obraz_skalowany + 127.5
kontrast_centrowany = np.clip(obraz_przesuniety_z_powrotem, 0, 255).astype(np.uint8)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(obraz_rgb)
plt.title("Oryginał")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(kontrast_centrowany)
plt.title(f"Kontrast centrowany (alpha={alpha_nowy_kontrast})")
plt.axis("off")
plt.show()
```

```python
szary = cv2.cvtColor(obraz, cv2.COLOR_BGR2GRAY)
plt.hist(szary.ravel(), bins=256, range=(0, 255))
plt.title("Histogram jasnosci")
plt.xlabel("Wartosc piksela")
plt.ylabel("Liczba pikseli")
plt.show()
```

```python
mglisty = cv2.imread("obraz_mglisty.jpg")
mglisty_szary = cv2.cvtColor(mglisty, cv2.COLOR_BGR2GRAY)
wyrownany = cv2.equalizeHist(mglisty_szary)
plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(mglisty_szary, cmap="gray")
plt.title("Przed wyrownaniem")
plt.axis("off")
plt.subplot(2, 2, 2)
plt.imshow(wyrownany, cmap="gray")
plt.title("Po wyrownaniu")
plt.axis("off")
plt.subplot(2, 2, 3)
plt.hist(mglisty_szary.ravel(), bins=256, range=(0, 255))
plt.title("Histogram przed")
plt.subplot(2, 2, 4)
plt.hist(wyrownany.ravel(), bins=256, range=(0, 255))
plt.title("Histogram po")
plt.show()
```

```python
maly = cv2.resize(obraz_rgb, (320, 240))
print("Wymiary przed:", obraz_rgb.shape)
print("Wymiary po:", maly.shape)
powiekszony = cv2.resize(obraz_rgb, None, fx=1.5, fy=1.5)
print("Wymiary po powiekszeniu:", powiekszony.shape)
```

```python
obrot90 = cv2.rotate(obraz_rgb, cv2.ROTATE_90_CLOCKWISE)
odbicie_poziome = cv2.flip(obraz_rgb, 1)
(wys, szer) = obraz_rgb.shape[:2]
srodek = (szer // 2, wys // 2)
macierz = cv2.getRotationMatrix2D(srodek, 30, 1.0)
obrot30 = cv2.warpAffine(obraz_rgb, macierz, (szer, wys))
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(obrot90)
plt.title("Obrot o 90 stopni")
plt.axis("off")
plt.subplot(1, 3, 2)
plt.imshow(odbicie_poziome)
plt.title("Odbicie poziome")
plt.axis("off")
plt.subplot(1, 3, 3)
plt.imshow(obrot30)
plt.title("Obrot o 30 stopni")
plt.axis("off")
plt.show()
```

# ZADANIE DODATKOWE

```python
obraz = cv2.imread("obraz_testowy.jpg")
obraz_rgb = cv2.cvtColor(obraz, cv2.COLOR_BGR2RGB)

wiekszy_kontrast = cv2.convertScaleAbs(obraz_rgb, alpha=1.6, beta=0)
mniejszy_kontrast = cv2.convertScaleAbs(obraz_rgb, alpha=0.5, beta=0)
negatyw = 255 - obraz_rgb

lewa = np.vstack((obraz, wiekszy_kontrast))
prawa = np.vstack((mniejszy_kontrast, negatyw))
obraz = np.hstack((lewa, prawa))

plt.imshow(obraz)
plt.title("Kolaz")
plt.axis("off")
plt.show()

cv2.imwrite("kolaz.png", cv2.cvtColor(obraz, cv2.COLOR_RGB2BGR))
```
